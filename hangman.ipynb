```python
import random

word=["JAZZ" ,"BUZZ" ,"PUPPY", "GHOST" , "PIZZA"]

words=random.choice(word)

chances= 6

while chances > 0:

    Guess=input("Enter The Word For The Correct Guess:")

    if Guess == word:
        print("Your Guess Is Correct")
    else:
        chances=chances-1
        print("Wrong Guess")
        print("Chances Left",chances)
    if Guess != words:
        print("Game Over")
        print("Try Again")
    
```
