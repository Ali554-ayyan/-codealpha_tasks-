```python
def Chatbot():
    user = input("Please Enter A Question: ").strip().lower()

    if user == "hello":
        print("Bot: Hi!")
    elif user == "how are you":
        print("Bot: I'm fine")
    elif user == "bye":
        print("Bot: Goodbye")
    else:
        print("Bot: I don't understand")

Chatbot()
```
